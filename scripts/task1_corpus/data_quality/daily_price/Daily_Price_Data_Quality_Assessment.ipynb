```python
import pandas as pd
```

```python
stocks_df = pd.read_csv('all_stocks_5yr.csv')
print(stocks_df.head(5))
```

```python
# Display dataset information
print(stocks_df.info())
```

```python
# Display total number of records
print("Total records:", len(stocks_df))
```

```python
# Check for missing values
stocks_df.isna().sum()
```

```python
# Check duplicate rows
print("Duplicate rows:", stocks_df.duplicated().sum())

# Check duplicate stock-date combinations
duplicates = stocks_df.duplicated(
    subset=["Name", "date"],
    keep=False
)
```

```python
price_columns = ["open", "high", "low", "close"]

for col in price_columns:
    print(f"Negative {col} prices:", (stocks_df[col] < 0).sum())
    print(f"Zero {col} prices:", (stocks_df[col] == 0).sum())

print("Negative volume:", (stocks_df["volume"] < 0).sum())
```

```python
# Check if high price is lower than low price
invalid_high_low = stocks_df[stocks_df["high"] < stocks_df["low"]]

# Check if high price is lower than open or close price
invalid_high = stocks_df[
    (stocks_df["high"] < stocks_df["open"]) |
    (stocks_df["high"] < stocks_df["close"])
]

# Check if low price is higher than open or close price
invalid_low = stocks_df[
    (stocks_df["low"] > stocks_df["open"]) |
    (stocks_df["low"] > stocks_df["close"])
]

# Display the number of invalid records
print("High < Low:", len(invalid_high_low))
print("High < Open/Close:", len(invalid_high))
print("Low > Open/Close:", len(invalid_low))
```

```python
# Convert date column to datetime
dates = pd.to_datetime(stocks_df["date"], errors="coerce")

# Check invalid dates
print("Invalid dates:", dates.isnull().sum())
```
